```sql
DESCRIBE TABLE workspace.gold.ibm_churn_overview;

DESCRIBE TABLE workspace.gold.ibm_churn_by_contract;

DESCRIBE TABLE workspace.gold.ibm_churn_by_payment_method;

DESCRIBE TABLE workspace.gold.ibm_churn_by_service;

DESCRIBE TABLE workspace.gold.ibm_churn_by_tenure;

DESCRIBE TABLE workspace.gold.trai_lsa_switching_usage_context;

DESCRIBE TABLE workspace.gold.trai_national_wireless_usage_trends;

SELECT *
FROM workspace.gold.ibm_churn_overview;

SELECT *
FROM workspace.gold.ibm_churn_by_contract;

SELECT *
FROM workspace.gold.ibm_churn_by_payment_method;

SELECT *
FROM workspace.gold.ibm_churn_by_service;

SELECT *
FROM workspace.gold.ibm_churn_by_tenure;

SELECT *
FROM workspace.gold.trai_lsa_switching_usage_context
LIMIT 20;

SELECT *
FROM workspace.gold.trai_national_wireless_usage_trends
LIMIT 20;


CREATE OR REPLACE VIEW workspace.gold.ibm_churn_segment_analysis AS

SELECT
    'Contract' AS segment_type,
    Contract AS segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_by_contract

UNION ALL

SELECT
    'Payment Method' AS segment_type,
    Payment_Method AS segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_by_payment_method

UNION ALL

SELECT
    'Internet Service' AS segment_type,
    Internet_Service AS segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_by_service

UNION ALL

SELECT
    'Customer Tenure' AS segment_type,
    tenure_band AS segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_by_tenure;


SELECT *
FROM workspace.gold.ibm_churn_segment_analysis
ORDER BY churn_rate_percent DESC;


SELECT
    segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_segment_analysis
WHERE segment_type = 'Contract'
ORDER BY churn_rate_percent DESC;


SELECT
    segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_segment_analysis
WHERE segment_type = 'Payment Method'
ORDER BY churn_rate_percent DESC;


SELECT
    segment,
    total_customers,
    churned_customers,
    churn_rate_percent
FROM workspace.gold.ibm_churn_segment_analysis
WHERE segment_type = 'Internet Service'
ORDER BY churn_rate_percent DESC;
```
