```sql
SHOW TABLES IN workspace.gold;

DESCRIBE TABLE workspace.gold.trai_lsa_switching_usage_context;

SELECT *
FROM workspace.gold.trai_lsa_switching_usage_context
LIMIT 10;

SELECT
    canonical_lsa,
    zone,
    may_2026_porting_requests_millions,
    june_2026_porting_requests_millions,
    ROUND(
        june_2026_porting_requests_millions -
        may_2026_porting_requests_millions, 3
    ) AS change_in_porting_requests,
    ROUND(
        (
            (june_2026_porting_requests_millions -
             may_2026_porting_requests_millions)
            / NULLIF(may_2026_porting_requests_millions, 0)
        ) * 100, 2
    ) AS percentage_change
FROM workspace.gold.trai_lsa_switching_usage_context
ORDER BY june_2026_porting_requests_millions DESC;


SELECT
    COUNT(DISTINCT canonical_lsa) AS total_lsa,
    COUNT(DISTINCT zone) AS total_zones,
    SUM(may_2026_porting_requests_millions)
        AS may_porting_requests,
    SUM(june_2026_porting_requests_millions)
        AS june_porting_requests,
    ROUND(AVG(prepaid_percentage), 2)
        AS avg_prepaid_percentage,
    ROUND(AVG(arpu_inr_per_month), 2)
        AS avg_arpu
FROM workspace.gold.trai_lsa_switching_usage_context;
```
