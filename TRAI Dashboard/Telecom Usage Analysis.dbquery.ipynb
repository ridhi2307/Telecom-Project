```sql
SELECT
    canonical_lsa,
    ROUND(AVG(outgoing_mou_per_month), 2)
        AS average_outgoing_mou,
    ROUND(AVG(incoming_mou_per_month), 2)
        AS average_incoming_mou,
    ROUND(AVG(total_mou_per_month), 2)
        AS average_total_mou
FROM workspace.gold.trai_lsa_switching_usage_context
GROUP BY canonical_lsa
ORDER BY average_total_mou DESC;
```
